#Import library

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import glob, os
import joblib
import kagglehub

from scipy.stats import randint, uniform
from sklearn.model_selection import GroupShuffleSplit, GroupKFold, RandomizedSearchCV
from sklearn.feature_selection import mutual_info_classif
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.utils.class_weight import compute_sample_weight
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             roc_auc_score, balanced_accuracy_score, classification_report,
                             ConfusionMatrixDisplay, RocCurveDisplay)

pd.set_option("display.max_columns", None)
sns.set_theme(style="whitegrid")
RANDOM_STATE = 42
````

#Donwload Dataset

````python
path = kagglehub.dataset_download("fedesoriano/gender-pay-gap-dataset")
print("Lokasi:", path)

for f in glob.glob(os.path.join(path, "**", "*"), recursive=True):
    if os.path.isfile(f):
        print(os.path.relpath(f, path), "-", round(os.path.getsize(f)/1024, 1), "KB")
````

#Membca file dataset

````python
def baca_csv(nama_file):
    lokasi = glob.glob(os.path.join(path, "**", nama_file), recursive=True)[0]
    return pd.read_csv(lokasi, low_memory=False)

df_cps  = baca_csv("CurrentPopulationSurvey.csv")
df_psid = baca_csv("PanelStudyIncomeDynamics.csv")

print("CPS :", df_cps.shape)
print("PSID:", df_psid.shape)
````

#Fungsi pengecekan (struktur, jumlah, kualitas)

````python
def cek_data(df, nama):
    print("=" * 70)
    print(f"  DATASET: {nama}")
    print("=" * 70)

    # --- Struktur data ---
    print("\n[STRUKTUR DATA]")
    display(df.head())
    print("Tipe data:")
    print(df.dtypes.value_counts().to_string())
    print(f"Memori: {df.memory_usage(deep=True).sum() / 1024**2:.1f} MB")

    # --- Jumlah data ---
    print("\n[JUMLAH DATA]")
    print(f"Jumlah baris : {df.shape[0]:,}")
    print(f"Jumlah kolom : {df.shape[1]:,}")

    # --- Kualitas data ---
    print("\n[KUALITAS DATA]")
    missing = df.isnull().sum()
    persen  = (missing / len(df) * 100).round(2)
    tabel_missing = (pd.DataFrame({"jumlah_missing": missing, "persen": persen})
                     .query("jumlah_missing > 0")
                     .sort_values("persen", ascending=False))
    print(f"Kolom dengan missing value : {len(tabel_missing)}")
    print(f"Kolom kosong total (100%)  : {(persen == 100).sum()}")
    print(f"Kolom missing > 50%        : {(persen > 50).sum()}")
    display(tabel_missing.head(20))

    print(f"Jumlah baris duplikat : {df.duplicated().sum():,}")

    konstan = [c for c in df.columns if df[c].nunique(dropna=True) <= 1]
    print(f"Kolom konstan ({len(konstan)}): {konstan}")

    print("\nKolom bertipe teks:")
    for c in df.select_dtypes("object").columns:
        print(f"  {c}: {df[c].nunique()} nilai unik, contoh → {df[c].dropna().unique()[:5]}")

    print("\nStatistik deskriptif:")
    display(df.describe().T)
````

#CPS

#PSID

````python
cek_data(df_psid, "Panel Study of Income Dynamics (PSID)")
````

#Bandingkan kolom kedua dataset

````python
kolom_cps, kolom_psid = set(df_cps.columns), set(df_psid.columns)

print(f"Kolom yang sama   : {len(kolom_cps & kolom_psid)}")
print(f"Hanya ada di CPS  : {len(kolom_cps - kolom_psid)}")
print(f"Hanya ada di PSID : {len(kolom_psid - kolom_cps)}")
print("\nKolom yang hanya ada di PSID:\n", sorted(kolom_psid - kolom_cps))
print("\nKolom yang hanya ada di CPS:\n", sorted(kolom_cps - kolom_psid))
````

#Cek dan bersihkan duplikat

````python
print("Duplikat CPS :", df_cps.duplicated().sum())
print("Duplikat PSID:", df_psid.duplicated().sum())

df_cps  = df_cps.drop_duplicates()
df_psid = df_psid.drop_duplicates()
````

#Seleksi kolom dan bersihkan missing value inti

````python
USE_UNION = False   # union dibuang: 83,7% baris CPS kosong (tidak ditanya) dan hubungannya dengan target lemah

CPS_FEATS = ['year','age','female','white','black','hisp','othrace',
             'sch','ba','adv','potexp','uhrswork','ft',
             'adj_ind2','adj_occ2','northeast','northcentral','south','west','metro']
PSID_FEATS = ['wave','age','female','white','black','hisp','othrace',
              'sch','ba','adv','potexp','usualhrwk','ft',
              'ind2','occ2','northeast','northcentral','south','west','smsa']
# Tidak dipakai:
# - annhrs, annwks, wkswork1 → division bias (upah per jam dihitung dari jam tahunan)
# - incwage, hrwage, lnrwg, dan turunan upah lain → data leakage
# - selfemp, military → konstan (0 semua) di sampel ini

if USE_UNION:
    CPS_FEATS.append('union')
    PSID_FEATS.append('unjob')

assert all(c in df_cps.columns  for c in CPS_FEATS),  "Ada kolom CPS yang tidak ditemukan"
assert all(c in df_psid.columns for c in PSID_FEATS), "Ada kolom PSID yang tidak ditemukan"

cps_clean  = df_cps[CPS_FEATS + ['realhrwage']].copy()
psid_clean = df_psid[PSID_FEATS + ['realhrwage', 'intnum68', 'pernum68']].copy()

# Bersihkan missing value inti: baris tanpa gender, usia, atau upah valid tidak bisa dipakai
for name, d in [('CPS', cps_clean), ('PSID', psid_clean)]:
    before = len(d)
    d.replace([np.inf, -np.inf], np.nan, inplace=True)
    d.dropna(subset=['female', 'age', 'realhrwage'], inplace=True)
    d.drop(d[d['realhrwage'] <= 0].index, inplace=True)   # upah nol/negatif tidak valid
    print(f"{name}: {before:,} -> {len(d):,} baris")

# Missing value yang tersisa di kolom fitur (untuk dokumentasi)
for name, d in [('CPS', cps_clean), ('PSID', psid_clean)]:
    m = d.drop(columns=['realhrwage']).isna().mean().mul(100).round(2)
    print(f"\nMissing value {name} (%):")
    print(m[m > 0].to_string() if (m > 0).any() else "tidak ada")

print("\nProporsi perempuan  CPS:", round(cps_clean['female'].mean(), 3),
      "| PSID:", round(psid_clean['female'].mean(), 3))
````

````python
nama_seragam_cps  = {'uhrswork':'jam_mingguan', 'adj_ind2':'industri', 'adj_occ2':'pekerjaan'}
nama_seragam_psid = {'wave':'year', 'unjob':'union', 'smsa':'metro', 'usualhrwk':'jam_mingguan',
                     'ind2':'industri', 'occ2':'pekerjaan'}

cps_g  = cps_clean.rename(columns=nama_seragam_cps)
psid_g = psid_clean.rename(columns=nama_seragam_psid)

# ID untuk split per orang: CPS = 1 baris 1 orang; PSID = ID keluarga + ID individu
cps_g['id_orang']  = 'C' + cps_g.index.astype(str)
psid_g['id_orang'] = 'P' + psid_g['intnum68'].astype(str) + '_' + psid_g['pernum68'].astype(str)
psid_g = psid_g.drop(columns=['intnum68', 'pernum68'])

cps_g['sumber_psid']  = 0
psid_g['sumber_psid'] = 1

df_gabung = pd.concat([cps_g, psid_g], ignore_index=True)
print("Gabungan:", df_gabung.shape)
print("Kolom yang tidak sama di kedua sumber:", sorted(set(cps_g.columns) ^ set(psid_g.columns)))
display(pd.crosstab(df_gabung['year'], df_gabung['sumber_psid']))
````

````python
kolom_cek = ['metro', 'industri', 'pekerjaan'] + (['union'] if USE_UNION else [])
for c in kolom_cek:
    print(f"\n===== {c} =====")
    display(pd.crosstab(df_gabung[c].fillna(-1), df_gabung['sumber_psid']))   # -1 = NaN
````

````python
cps = df_gabung['sumber_psid'] == 0

# metro: CPS 1 = non-metro, 2-4 = metro; PSID (smsa) sudah 0/1
if df_gabung.loc[cps, 'metro'].max() > 1:      # pengaman: tidak dobel jalan
    df_gabung.loc[cps, 'metro'] = df_gabung.loc[cps, 'metro'].map({0: np.nan, 1: 0, 2: 1, 3: 1, 4: 1})

# union (hanya kalau USE_UNION = True): 0 = tidak ditanya, 1 = tidak tercakup, 2 = anggota, 3 = tercakup tapi bukan anggota
if USE_UNION and df_gabung.loc[cps, 'union'].max() > 1:
    df_gabung.loc[cps, 'union'] = df_gabung.loc[cps, 'union'].map({0: np.nan, 1: 0, 2: 1, 3: 1})

for c in kolom_cek:
    display(pd.crosstab(df_gabung[c].fillna(-1), df_gabung['sumber_psid']))
````

````python
# Jam kerja mingguan > 168 bukan jam sungguhan (mis. kode 997/999 di IPUMS-CPS)
print("jam_mingguan tidak wajar:", (df_gabung['jam_mingguan'] > 168).sum())
df_gabung.loc[df_gabung['jam_mingguan'] > 168, 'jam_mingguan'] = np.nan

fitur = [c for c in df_gabung.columns if c not in ['realhrwage', 'id_orang']]

ringkas = pd.DataFrame({
    'tipe'         : df_gabung[fitur].dtypes.astype(str),
    'n_unik'       : df_gabung[fitur].nunique(),
    'min'          : df_gabung[fitur].min(numeric_only=True),
    'max'          : df_gabung[fitur].max(numeric_only=True),
    'missing_CPS%' : df_gabung.loc[df_gabung.sumber_psid == 0, fitur].isna().mean().mul(100).round(1),
    'missing_PSID%': df_gabung.loc[df_gabung.sumber_psid == 1, fitur].isna().mean().mul(100).round(1),
})
display(ringkas)

# Kolom konstan (mis. selfemp/military bernilai 0 semua) → dibuang
konstan = [c for c in fitur if df_gabung[c].nunique(dropna=True) <= 1]
print("Kolom konstan dibuang:", konstan)
df_gabung = df_gabung.drop(columns=konstan)
````

````python
METODE_AMBANG = 'kohavi'     # 'kohavi' | 'median' | 'dua_pertiga_median'

ambang_per_jam_2010 = 32.79  # Nilai ambang batas Kohavi dolar 2010

if METODE_AMBANG == 'median':
    ambang = df_gabung['realhrwage'].median()
elif METODE_AMBANG == 'dua_pertiga_median':          # definisi low pay OECD
    ambang = (2 / 3) * df_gabung['realhrwage'].median()
else:
    ambang = ambang_per_jam_2010                      # Kohavi (sel di atas)

df_gabung['upah_tinggi'] = (df_gabung['realhrwage'] > ambang).astype(int)

print(f"Metode: {METODE_AMBANG} | ambang = ${ambang:.2f}/jam (dolar 2010)")
print(df_gabung['upah_tinggi'].value_counts(normalize=True).round(3))
print(pd.crosstab(df_gabung['female'], df_gabung['upah_tinggi'], normalize='index').round(3))
````

````python
PAKAI_IND_OCC = True     # True = dengan industri & pekerjaan, False = tanpa (kesenjangan keseluruhan)
KOLOM_KATEGORI = ['metro'] + (['union'] if USE_UNION else []) + (['industri', 'pekerjaan'] if PAKAI_IND_OCC else [])

fitur_mentah = df_gabung.drop(columns=['realhrwage', 'id_orang', 'upah_tinggi']
                              + ([] if PAKAI_IND_OCC else ['industri', 'pekerjaan']))

def encode(df):
    d = df.copy()
    for c in KOLOM_KATEGORI:
        d[c] = d[c].apply(lambda v: 'nan' if pd.isna(v) else str(int(v)))
    return pd.get_dummies(d, columns=KOLOM_KATEGORI, dtype=int)

X_all = encode(fitur_mentah)
y_all = df_gabung['upah_tinggi']
print("Jumlah fitur:", X_all.shape[1])
print("Kolom:", X_all.columns.tolist())
````

````python
kolom_heat = [c for c in X_all.columns if not c.startswith(('industri_', 'pekerjaan_'))]
korelasi = X_all[kolom_heat].assign(upah_tinggi=y_all.values).corr()

plt.figure(figsize=(14, 11))
sns.heatmap(korelasi, cmap='coolwarm', center=0, linewidths=0.2)
plt.title('Heatmap Korelasi Fitur dan Target (tanpa dummy industri/pekerjaan)')
plt.tight_layout()
plt.show()

r_target = X_all.corrwith(y_all).sort_values()
top = r_target.reindex(r_target.abs().sort_values(ascending=False).index[:20]).sort_values()
plt.figure(figsize=(8, 7))
plt.barh(top.index, top.values, color=['#d62728' if v < 0 else '#1f77b4' for v in top.values])
plt.axvline(0, color='black', linewidth=0.8)
plt.title('20 Fitur dengan Korelasi Terkuat terhadap Upah Tinggi')
plt.tight_layout()
plt.show()
print("Korelasi female dengan target:", round(r_target['female'], 3))
````

````python
edu = np.select([X_all['adv'] == 1, X_all['ba'] == 1, X_all['sch'] <= 12],
                ['Pascasarjana', 'Sarjana', 'SMA ke bawah'], default='Some college')
tabel_edu = (pd.DataFrame({'pendidikan': edu, 'female': X_all['female'], 'upah_tinggi': y_all})
             .groupby(['pendidikan', 'female'])['upah_tinggi'].mean().unstack() * 100)
tabel_edu.columns = ['Laki-laki (%)', 'Perempuan (%)']
tabel_edu['Selisih (poin %)'] = tabel_edu['Laki-laki (%)'] - tabel_edu['Perempuan (%)']
display(tabel_edu.round(2))
````

````python
from scipy.stats import chi2_contingency

fitur_raw = fitur_mentah.columns.tolist()
y = df_gabung['upah_tinggi']

def cramers_v(x, y):
    tabel = pd.crosstab(x.fillna(-1), y)
    chi2 = chi2_contingency(tabel)[0]
    return np.sqrt(chi2 / (tabel.values.sum() * (min(tabel.shape) - 1)))

sampel = df_gabung.sample(50_000, random_state=RANDOM_STATE)
baris = []
for c in fitur_raw:
    kategori = c in KOLOM_KATEGORI
    kekuatan = cramers_v(df_gabung[c], y) if kategori else abs(df_gabung[c].corr(y))
    x_s = sampel[c].fillna(-1 if kategori else sampel[c].median())
    mi = mutual_info_classif(x_s.to_frame(), sampel['upah_tinggi'],
                             discrete_features=[kategori or x_s.nunique() <= 2],
                             random_state=RANDOM_STATE)[0]
    baris.append({'fitur': c,
                  'jenis': 'kategori (Cramér V)' if kategori else 'numerik/biner (|r|)',
                  'kekuatan': kekuatan, 'mi': mi,
                  'missing_%': df_gabung[c].isna().mean() * 100})

audit = pd.DataFrame(baris).set_index('fitur').sort_values('kekuatan', ascending=False)
audit['status'] = np.where((audit['kekuatan'] < 0.02) & (audit['mi'] < 0.001), 'lemah', 'ok')
display(audit.round(4))

# 1) Terlalu kuat = curiga leakage
print("Kekuatan > 0.5:", audit.index[audit['kekuatan'] > 0.5].tolist())

# 2) Redundansi antar-fitur numerik
num = [c for c in fitur_raw if c not in KOLOM_KATEGORI]
k = df_gabung[num].corr().abs()
pasang = [(a, b, round(k.loc[a, b], 3)) for i, a in enumerate(num) for b in num[i + 1:] if k.loc[a, b] > 0.8]
print("Pasangan fitur |r| > 0.8:", pasang)

# 3) Efek gender stabil di tiap sumber dan tahun?
for kunci in ['sumber_psid', 'year']:
    display(df_gabung.groupby(kunci)[['female', 'upah_tinggi']]
            .apply(lambda d: d['female'].corr(d['upah_tinggi'])).round(3).to_frame('r_female'))
````

````python
PROTECT = ['female', 'sumber_psid']

sampel = X_all.sample(50_000, random_state=RANDOM_STATE)
sampel_isi = sampel.fillna(sampel.median())
mi = pd.Series(
    mutual_info_classif(sampel_isi, y_all.loc[sampel.index],
                        discrete_features=[sampel_isi[c].nunique() <= 2 for c in sampel_isi.columns],
                        random_state=RANDOM_STATE),
    index=sampel_isi.columns)

r_abs = X_all.corrwith(y_all).abs().fillna(0)
lemah = [c for c in X_all.columns if r_abs[c] < 0.01 and mi[c] < 0.001 and c not in PROTECT]
print(f"Fitur dihapus ({len(lemah)}):", lemah)

X_sel = X_all.drop(columns=lemah)
print("Jumlah fitur akhir:", X_sel.shape[1])
````

````python
grup = df_gabung['id_orang']
gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=RANDOM_STATE)
idx_train, idx_test = next(gss.split(X_sel, y_all, groups=grup))

X_train, X_test = X_sel.iloc[idx_train].copy(), X_sel.iloc[idx_test].copy()
y_train, y_test = y_all.iloc[idx_train], y_all.iloc[idx_test]
grup_train = grup.iloc[idx_train]

# Sisa NaN (kolom numerik) diisi median TRAIN; GradientBoostingClassifier tidak menerima NaN
median_train = X_train.median()
X_train = X_train.fillna(median_train)
X_test  = X_test.fillna(median_train)

print("Train:", X_train.shape, "| Test:", X_test.shape)
print("Orang di train & test sekaligus:", len(set(grup_train) & set(grup.iloc[idx_test])))  # harus 0
print("NaN tersisa:", int(X_train.isna().sum().sum()), int(X_test.isna().sum().sum()))
````

````python
sw = compute_sample_weight('balanced', y_train)     # menangani ketidakseimbangan kelas
TUNING = False                                       # True = cari hyperparameter (lebih lama)

if TUNING:
    sub = np.random.RandomState(RANDOM_STATE).choice(len(X_train), 60_000, replace=False)
    Xs, ys, gs = X_train.iloc[sub], y_train.iloc[sub], grup_train.iloc[sub]
    search = RandomizedSearchCV(
        GradientBoostingClassifier(subsample=0.8, random_state=RANDOM_STATE),
        param_distributions={'n_estimators': randint(100, 400), 'learning_rate': uniform(0.03, 0.17),
                             'max_depth': randint(2, 6), 'min_samples_leaf': randint(5, 100)},
        n_iter=10, cv=GroupKFold(n_splits=3), scoring='f1', n_jobs=-1,
        random_state=RANDOM_STATE, verbose=1)
    search.fit(Xs, ys, groups=gs, sample_weight=compute_sample_weight('balanced', ys))
    print("Parameter terbaik:", search.best_params_, "| F1 CV:", round(search.best_score_, 4))
    gb = GradientBoostingClassifier(**search.best_params_, subsample=0.8, random_state=RANDOM_STATE)
else:
    gb = GradientBoostingClassifier(n_estimators=200, learning_rate=0.1, max_depth=4,
                                    subsample=0.8, random_state=RANDOM_STATE)

gb.fit(X_train, y_train, sample_weight=sw)

y_pred_train = gb.predict(X_train)
y_pred_test  = gb.predict(X_test)
y_prob_test  = gb.predict_proba(X_test)[:, 1]
````

````python
print("Proporsi kelas train:", y_train.value_counts(normalize=True).round(3).to_dict())
print("Proporsi kelas test :", y_test.value_counts(normalize=True).round(3).to_dict())

print("\nBalanced accuracy test:", round(balanced_accuracy_score(y_test, y_pred_test), 4))
print("Recall kelas rendah (0):", round(recall_score(y_test, y_pred_test, pos_label=0), 4))
print("Recall kelas tinggi (1):", round(recall_score(y_test, y_pred_test, pos_label=1), 4))

for nama, yt, yp in [('Train', y_train, y_pred_train), ('Test', y_test, y_pred_test)]:
    print(f"{nama}: accuracy={accuracy_score(yt, yp):.4f} | F1={f1_score(yt, yp):.4f}")
````

````python
print("Accuracy :", round(accuracy_score(y_test, y_pred_test), 4))
print("Precision:", round(precision_score(y_test, y_pred_test), 4))
print("Recall   :", round(recall_score(y_test, y_pred_test), 4))
print("F1-score :", round(f1_score(y_test, y_pred_test), 4))
print("ROC-AUC  :", round(roc_auc_score(y_test, y_prob_test), 4))
print("\n", classification_report(y_test, y_pred_test, target_names=['Upah rendah', 'Upah tinggi']))

fig, ax = plt.subplots(1, 2, figsize=(12, 5))
ConfusionMatrixDisplay.from_predictions(y_test, y_pred_test, display_labels=['Rendah', 'Tinggi'], cmap='Blues_r', ax=ax[0])
ax[0].set_title('Confusion Matrix')
RocCurveDisplay.from_predictions(y_test, y_prob_test, ax=ax[1])
ax[1].set_title('Kurva ROC')
plt.tight_layout()
plt.show()
````

````python
# @title Masukkan Profil Tenaga Kerja

age = 19 # @param {type:"slider", min:16, max:85, step:1}
female = "Laki-laki" # @param ["Laki-laki", "Perempuan"]
pendidikan = "Sarjana" # @param ["Pascasarjana", "Sarjana", "Some college", "SMA ke bawah"]
jam_mingguan = 31 # @param {type:"number"}
ft = "Part-time" # @param ["Full-time", "Part-time"]
region = "Northeast" # @param ["Northeast", "Northcentral", "South", "West"]
metro = "Metro" # @param ["Metro", "Non-Metro"]

# --- Ekstraksi Fitur Tambahan ---
female_val = 1 if female == "Perempuan" else 0
ft_val = 1 if ft == "Full-time" else 0

sch = 12
ba = 0
adv = 0
if pendidikan == "Pascasarjana":
    sch = 18
    ba = 1
    adv = 1
elif pendidikan == "Sarjana":
    sch = 16
    ba = 1
elif pendidikan == "Some college":
    sch = 14

potexp = max(0, age - sch - 6)

# --- Wilayah ---
ne, nc, s, w = 0, 0, 0, 0
if region == "Northeast": ne = 1
elif region == "Northcentral": nc = 1
elif region == "South": s = 1
elif region == "West": w = 1

# --- Pembuatan Dataframe Input (Inference) ---
input_data = pd.DataFrame(0, index=[0], columns=X_sel.columns)

# Isi nilai numerik/biner dasar
input_data['year'] = 2011  # Tahun acuan terakhir
input_data['age'] = age
input_data['female'] = female_val
input_data['white'] = 1     # Nilai baseline dominan
input_data['black'] = 0
input_data['hisp'] = 0
input_data['othrace'] = 0
input_data['sch'] = sch
input_data['ba'] = ba
input_data['adv'] = adv
input_data['potexp'] = potexp
input_data['jam_mingguan'] = jam_mingguan
input_data['ft'] = ft_val
input_data['northeast'] = ne
input_data['northcentral'] = nc
input_data['south'] = s
input_data['west'] = w
input_data['sumber_psid'] = 0  # Default menggunakan bobot CPS

# Mengisi kolom metro
if metro == "Metro":
    if 'metro_1' in input_data.columns: input_data['metro_1'] = 1
else:
    if 'metro_0' in input_data.columns: input_data['metro_0'] = 1

# Lakukan prediksi
proba = gb.predict_proba(input_data)[0][1]
prediksi = gb.predict(input_data)[0]

# Tampilkan hasil prediksi dengan visualisasi sederhana
print("=" * 50)
print("               HASIL INFERENSI MODEL")
print("=" * 50)
print(f"Profil Input     : {female}, {age} tahun, Pendidikan {pendidikan}")
print(f"Jam Kerja        : {jam_mingguan} jam/minggu ({ft})")
print(f"Wilayah          : {region} ({metro})")
print("-" * 50)
print(f"Probabilitas Upah Tinggi (> $32.79/jam): {proba * 100:.2f}%")
if prediksi == 1:
    print("Klasifikasi Model: UPAH TINGGI [1]")
else:
    print("Klasifikasi Model: UPAH RENDAH [0]")
print("=" * 50)
````
